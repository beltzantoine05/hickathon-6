```python
import pandas as pd

X_train = pd.read_csv('X_train.csv', sep = ',', index_col=0)
X_test = pd.read_csv('X_test.csv', sep = ',', index_col=0)

for c in ['PA008', 'ST006', 'ST008']:
    X_train[c] = X_train[c].fillna(X_train[f"{c}.1"])
    X_test[c] = X_test[c].fillna(X_train[f"{c}.1"])

X_train.drop(columns=['PA008.1', 'ST006.1', 'ST008.1'], inplace=True)
X_test.drop(columns=['PA008.1', 'ST006.1', 'ST008.1'], inplace=True)

X_train.to_csv("X_train.csv")
X_test.to_csv("X_test.csv")

#X_train["D_INDEX"] = X_train["CNTSTUID"].astype(str) + "/" + X_train["Year"].astype(str)
```

```python
# 1) nb de tentatives par (étudiant, année)
attempts = X_train.groupby(["CNTSTUID", "Year"]).size()

# 2) cumsum par étudiant
cum = attempts.groupby(level=0).cumsum()

# 3) TIME = tentatives avant cette année + 1
time_per_year = (cum - attempts + 1).rename("TIME")

# 4) join sur les colonnes CNTSTUID, Year
X_train = X_train.join(time_per_year, on=["CNTSTUID", "Year"])
```

```python
import torch
from sklearn.preprocessing import RobustScaler
from sklearn.impute import SimpleImputer
from sklearn.model_selection import KFold
import numpy as np

kfold = KFold(n_splits=5, shuffle=True, random_state=42)
for fold, (train_idx, val_idx) in enumerate(kfold.split(data_raw)):
    X_train_raw = data_raw.iloc[train_idx]
    X_val_raw = data_raw.iloc[val_idx]

    imputer = SimpleImputer(strategy="mean", keep_empty_features=True)
    X_train_imputed = imputer.fit_transform(X_train_raw)

    scaler = RobustScaler()
    X_train_scaled = scaler.fit_transform(X_train_imputed).astype(np.float32)

    if X_val_raw is not None:
        mask_val = (~np.isnan(X_val_raw)).astype(np.float32)
        X_val_imputed = imputer.transform(X_val_raw)
        X_val_scaled = scaler.transform(X_val_imputed).astype(np.float32)
    else:
        val_loader = None

    obs_train = X_train_scaled
    obs_val = X_val_scaled
    print("train |x| p50/p90/p99/max:", np.quantile(np.abs(obs_train), [0.5,0.9,0.99]), np.max(np.abs(obs_train)))
    print("val   |x| p50/p90/p99/max:", np.quantile(np.abs(obs_val), [0.5,0.9,0.99]), np.max(np.abs(obs_val)))

    print("scaler.scale_ min/p1/p5/median:", np.min(scaler.scale_), np.quantile(scaler.scale_, 0.01),
      np.quantile(scaler.scale_, 0.05), np.median(scaler.scale_))

    max_abs = np.max(np.abs(X_train_scaled), axis=0)
    n_obs = len(train_idx)
    df_dbg = pd.DataFrame({
        "col": columns_dae_piza,
        "max_abs": max_abs,
        "n_obs": n_obs,
        "scale": scaler.scale_,
    }).sort_values("max_abs", ascending=False)
    print(df_dbg.head(20))

```
