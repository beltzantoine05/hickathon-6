```python
X_train = X_train.drop(columns=cols_to_drop)
exo_embedding_finetuned_full_train = pd.read_csv("data/Embedding_wandb/exo_embedding_finetuned_full_train.csv",index_col=0)
que_embedding_finetuned_full_train = pd.read_csv("data/Embedding_wandb/que_embedding_finetuned_full_train.csv",index_col=0)
df_preprocessed = pd.read_csv("data/df_processed.csv",index_col=0)
mask = df_preprocessed["MathScore"].notna()
df_preprocessed= df_preprocessed.loc[mask]

X_train = X_train.join(exo_embedding_finetuned_full_train, how='left')
X_train = X_train.join(que_embedding_finetuned_full_train, how='left')

# X_train["STRATUM_CLUST"] =df_preprocessed["STRATUM"]
# X_train["DIFFERENT"] =df_preprocessed["DIFFERENT"]
# X_train["OCOD1_ISEI"] =df_preprocessed["OCOD1"]
# X_train["OCOD2_ISEI"] =df_preprocessed["OCOD2"]
# X_train["OCOD3_ISEI"] =df_preprocessed["OCOD3"]
# X_train["IND_ST001D01T"] =df_preprocessed["ST001D01T"]
```

```python
X_train["MathScore"] = df_preprocessed["MathScore"]
```

```python
import pandas as pd
from autogluon.tabular import TabularDataset, TabularPredictor

train_data = TabularDataset(X_train)
train_data.head()

```

```python
label = 'MathScore'
time_limit = 105*60
metric = 'r2'
predictor = TabularPredictor(label, eval_metric=metric).fit(train_data,
                                                            time_limit=time_limit,
                                                              presets='extreme_quality', num_gpus=1,
                                                              dynamic_stacking=False, num_stack_levels=1,
                                                              excluded_model_types=['RF'])
```

```python
X_test = X_test.drop(columns=cols_to_drop)
exo_embedding_finetuned_full_test = pd.read_csv("data/Embedding_wandb/exo_embedding_finetuned_full_test.csv",index_col=0)
que_embedding_finetuned_full_test = pd.read_csv("data/Embedding_wandb/que_embedding_finetuned_full_test.csv",index_col=0)
df_preprocessed_test = pd.read_csv("data/df_processed.csv",index_col=0)
mask_test = df_preprocessed_test["MathScore"].isna()
df_preprocessed_test= df_preprocessed_test.loc[mask_test]

X_test = X_test.join(exo_embedding_finetuned_full_test, how='left')
X_test = X_test.join(que_embedding_finetuned_full_test, how='left')

# X_test["STRATUM_CLUST"] =df_preprocessed_test["STRATUM"]
# X_test["DIFFERENT"] =df_preprocessed_test["DIFFERENT"]
# X_test["OCOD1_ISEI"] =df_preprocessed_test["OCOD1"]
# X_test["OCOD2_ISEI"] =df_preprocessed_test["OCOD2"]
# X_test["OCOD3_ISEI"] =df_preprocessed_test["OCOD3"]
# X_test["IND_ST001D01T"] =df_preprocessed_test["ST001D01T"]
```

```python
test_data = TabularDataset(X_test)
test_data.head()
```

```python
y_pred = predictor.predict(test_data)
y_pred.head()
```

```python
y_pred = y_pred.clip(lower=0, upper=900)
```
