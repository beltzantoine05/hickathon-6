```python
%%bash
set -euo pipefail

# --- Réglages ---
ENV_NAME="ag"
PY_VER="3.12"
KERNEL_NAME="ag"
# Nom affiché dans Jupyter
KERNEL_DISPLAY="Python (${ENV_NAME}-${PY_VER} + AutoGluon)"
# Chemin pour stocker les environnements Conda (dans $HOME pour l'isolation)
ROOT_DIR="$HOME/micromamba-root"
# Chemin pour l'exécutable micromamba (dans le répertoire courant)
BIN_DIR="$PWD/bin"
MAMBA="${BIN_DIR}/micromamba"

echo "=== 1) Téléchargement micromamba (si absent) ==="
mkdir -p "$BIN_DIR"
if [[ ! -x "$MAMBA" ]]; then
    echo "Téléchargement de l'exécutable micromamba..."
    # Téléchargement : curl | tar (plus rapide)
    curl -Ls https://micro.mamba.pm/api/micromamba/linux-64/latest \
      | tar -xvj -C "$BIN_DIR" bin/micromamba
fi
# Test de l'exécutable (doit s'exécuter sans erreur)
"$MAMBA" --help >/dev/null

# Important : Assurer que les environnements Conda sont stockés dans le ROOT_DIR isolé.
export MAMBA_ROOT_PREFIX="$ROOT_DIR"
echo "MAMBA_ROOT_PREFIX défini sur: $ROOT_DIR"

echo "=== 2) Création de l'environnement Python ${PY_VER} ==="
# Vérifie si l'environnement existe déjà pour éviter une recréation inutile
if ! "$MAMBA" env list | awk '{print $1}' | grep -qx "${ENV_NAME}"; then
    echo "Création de l'environnement '${ENV_NAME}' avec Python ${PY_VER} et pip..."
    "$MAMBA" create -y -n "$ENV_NAME" "python=${PY_VER}" pip
else
    echo "Env '${ENV_NAME}' existe déjà, on le met à jour."
fi

echo "=== 3) Installation de AutoGluon et ipykernel ==="
# Mise à jour pip pour assurer la compatibilité
"$MAMBA" run -n "$ENV_NAME" python -m pip install --upgrade pip

# Installation d'AutoGluon avec les extras recommandés (tabular et vision)
# J'ai ajouté un tilde (~) pour la version 1.1 pour assurer les mises à jour mineures
echo "Installation de autogluon[tabular,vision]~=1.1..."
"$MAMBA" run -n "$ENV_NAME" pip install 'autogluon[tabular,vision]~=1.1'

# Installation d'ipykernel pour l'utiliser dans Jupyter
"$MAMBA" run -n "$ENV_NAME" pip install ipykernel

echo "=== 4) Enregistrement du kernel Jupyter ==="
# Enregistre l'environnement comme un kernel Jupyter accessible par l'utilisateur
"$MAMBA" run -n "$ENV_NAME" python -m ipykernel install --user \
  --name "$KERNEL_NAME" --display-name "$KERNEL_DISPLAY"

echo "=== 5) Vérification rapide ==="
# Exécute un petit script Python dans le nouvel environnement
"$MAMBA" run -n "$ENV_NAME" python - <<'PY'
import sys, pkgutil
print("OK -> Python:", sys.version.split()[0])
try:
    import importlib.metadata as im
except ImportError:
    import importlib_metadata as im
try:
    v = im.version("autogluon")
    print("AutoGluon (Version):", v)
except Exception as e:
    pkgs = sorted(m.name for m in pkgutil.iter_modules() if m.name.startswith("autogluon"))
    print("Impossible de lire la version via metadata:", e)
    print("Sous-paquets détectés:", pkgs)
PY

echo
echo "=============================================================="
echo "Terminé 🎉 L'environnement est prêt."
echo "Pour l'utiliser dans ce notebook : Kernel -> Change kernel -> \"${KERNEL_DISPLAY}\""
echo "=============================================================="
```

```python
X_train = X_train.drop(columns=cols_to_drop)
exo_embedding_finetuned_full_train = pd.read_csv("data/Embedding_wandb/exo_embedding_finetuned_full_train.csv",index_col=0,nrows=1000)
que_embedding_finetuned_full_train = pd.read_csv("data/Embedding_wandb/que_embedding_finetuned_full_train.csv",index_col=0,nrows=1000)
df_preprocessed = pd.read_csv("data/df_processed.csv",index_col=0,nrows=1000)
mask = df_preprocessed["MathScore"].notna()
df_preprocessed= df_preprocessed.loc[mask]

X_train = X_train.join(exo_embedding_finetuned_full_train, how='left')
X_train = X_train.join(que_embedding_finetuned_full_train, how='left')

# X_train["STRATUM_CLUST"] =df_preprocessed["STRATUM"]
# X_train["DIFFERENT"] =df_preprocessed["DIFFERENT"]
# X_train["OCOD1_ISEI"] =df_preprocessed["OCOD1"]
# X_train["OCOD2_ISEI"] =df_preprocessed["OCOD2"]
# X_train["OCOD3_ISEI"] =df_preprocessed["OCOD3"]
# X_train["IND_ST001D01T"] =df_preprocessed["ST001D01T"]
```

```python
X_train["MathScore"] = df_preprocessed["MathScore"]
```

```python
import pandas as pd
from autogluon.tabular import TabularDataset, TabularPredictor

train_data = TabularDataset(X_train)
train_data.head()

```

```python
label = 'MathScore'
time_limit = 10*60
metric = 'r2'
predictor = TabularPredictor(label, eval_metric=metric).fit(train_data,
                                                            time_limit=time_limit,
                                                              presets='medium_quality', num_gpus=1,
                                                              dynamic_stacking=False, num_stack_levels=1,
                                                              excluded_model_types=['RF'])
```
